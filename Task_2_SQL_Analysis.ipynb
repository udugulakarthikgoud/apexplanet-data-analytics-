```python
query = """
SELECT
    s."Customer ID",
    c."Customer Name",
    c.Segment,
    SUM(s.Sales) AS Total_Sales
FROM sales s
JOIN customers c
    ON s."Customer ID" = c."Customer ID"
GROUP BY
    s."Customer ID",
    c."Customer Name",
    c.Segment
ORDER BY Total_Sales DESC
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT "Product Name", Sales
FROM sales
WHERE Sales > (
    SELECT AVG(Sales)
    FROM sales
)
ORDER BY Sales DESC
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
WITH category_sales AS (
    SELECT
        Category,
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Category
)
SELECT
    Category,
    Total_Sales
FROM category_sales
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT "Product Name", Sales
FROM sales
WHERE Sales > (
    SELECT AVG(Sales)
    FROM sales
)
ORDER BY Sales DESC
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT
    "Customer Name",
    Sales,
    RANK() OVER (
        ORDER BY Sales DESC
    ) AS Sales_Rank
FROM sales
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT
    "Customer Name",
    Sales,
    LAG(Sales) OVER (
        ORDER BY Sales
    ) AS Previous_Sales
FROM sales
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT
    "Customer Name",
    Sales,
    LEAD(Sales) OVER (
        ORDER BY Sales
    ) AS Next_Sales
FROM sales
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
CREATE VIEW IF NOT EXISTS category_sales AS
SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Category;
"""

conn.execute(query)
conn.commit()

print("View created successfully!")
```

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("sqlite:///superstore.db")

print("Pandas and SQLAlchemy are ready!")
```

```python
query = """
SELECT
    strftime('%Y-%m', "Order Date") AS Month,
    SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Month
ORDER BY Month;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT
    "Customer ID",
    "Customer Name",
    SUM(Sales) AS Total_Spend,
    CASE
        WHEN SUM(Sales) >= 5000 THEN 'High'
        WHEN SUM(Sales) >= 2000 THEN 'Medium'
        ELSE 'Low'
    END AS Spend_Category
FROM sales
GROUP BY "Customer ID", "Customer Name"
ORDER BY Total_Spend DESC;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT
    Region,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Region
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Category
ORDER BY Total_Profit DESC;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT
    "Customer ID",
    "Customer Name",
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Customer ID", "Customer Name"
ORDER BY Total_Profit DESC
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT
    "Sub-Category",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Sub-Category"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT
    Region,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(SUM(Profit) * 100.0 / SUM(Sales), 2) AS Profit_Margin_Percent
FROM sales
GROUP BY Region
ORDER BY Profit_Margin_Percent DESC;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT
    "Ship Mode",
    COUNT(*) AS Total_Orders,
    SUM(Sales) AS Total_Sales
FROM sales
GROUP BY "Ship Mode"
ORDER BY Total_Orders DESC;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
SELECT
    State,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY State
ORDER BY Total_Sales DESC
LIMIT 10;
"""

result = pd.read_sql(query, engine)
result
```
