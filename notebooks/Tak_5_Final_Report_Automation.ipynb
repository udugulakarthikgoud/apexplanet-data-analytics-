```python
import pandas as pd
import numpy as np
import os
```

```python
# Load the cleaned dataset
data = pd.read_csv("Superstore_final_cleaned.csv")

print("Dataset loaded successfully!")
print("Shape:", data.shape)
data.head()
```

```python
# Check dataset information
print("Dataset Information")
print("=" * 40)

print("Rows and Columns:", data.shape)
print("\nColumn Names:")
print(data.columns.tolist())

print("\nMissing Values:")
print(data.isnull().sum().sum())

print("\nDuplicate Rows:")
print(data.duplicated().sum())
```

```python
# Calculate Key Performance Indicators (KPIs)

total_sales = data["Sales"].sum()
total_profit = data["Profit"].sum()
total_quantity = data["Quantity"].sum()
total_orders = data["Order ID"].nunique()

print("KEY PERFORMANCE INDICATORS")
print("=" * 40)
print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity:", total_quantity)
print("Total Orders:", total_orders)
```

```python
# Create KPI summary
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Profit",
        "Total Quantity",
        "Total Orders"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_quantity,
        total_orders
    ]
})

# Export KPIs to Excel
kpi_summary.to_excel("Task_5_KPI_Report.xlsx", index=False)

print("KPI report exported successfully!")
```

```python
# Automated Data Cleaning

raw_data = pd.read_csv("Superstore sales dataset.csv")

# Remove duplicate rows
processed_data = raw_data.drop_duplicates()

# Remove rows with missing values
processed_data = processed_data.dropna()

# Save processed data
processed_data.to_csv("Superstore_processed.csv", index=False)

print("Data cleaning completed successfully!")
print("Original rows:", len(raw_data))
print("Processed rows:", len(processed_data))
print("Processed dataset saved as: Superstore_processed.csv")
```

```python
# Export processed data to Excel

processed_data.to_excel("Superstore_processed.xlsx", index=False)

print("Processed data exported successfully!")
print("File: Superstore_processed.xlsx")
```
