```python

# Descriptive Statistics

print("DESCRIPTIVE STATISTICS")
print("=" * 40)

print("\nSALES")
print("Mean:", df["Sales"].mean())
print("Median:", df["Sales"].median())
print("Mode:", df["Sales"].mode()[0])
print("Standard Deviation:", df["Sales"].std())
print("Skewness:", df["Sales"].skew())

print("\nPROFIT")
print("Mean:", df["Profit"].mean())
print("Median:", df["Profit"].median())
print("Mode:", df["Profit"].mode()[0])
print("Standard Deviation:", df["Profit"].std())
print("Skewness:", df["Profit"].skew())
```

```python
# Statistical Summary

print("STATISTICAL SUMMARY")
print("=" * 40)

df[["Sales", "Quantity", "Discount", "Profit"]].describe()
```

```python
df[["Sales", "Quantity", "Discount", "Profit"]].describe()
```

```python
# Distribution Analysis

print("SKEWNESS ANALYSIS")
print("=" * 40)

print("Sales Skewness:", df["Sales"].skew())
print("Profit Skewness:", df["Profit"].skew())

if df["Sales"].skew() > 0:
    print("Sales: Positively Skewed")
elif df["Sales"].skew() < 0:
    print("Sales: Negatively Skewed")
else:
    print("Sales: Approximately Symmetric")

if df["Profit"].skew() > 0:
    print("Profit: Positively Skewed")
elif df["Profit"].skew() < 0:
    print("Profit: Negatively Skewed")
else:
    print("Profit: Approximately Symmetric")
```

```python
# Hypothesis Testing - Independent t-test

from scipy.stats import ttest_ind

# Compare Profit between two categories
category1 = df[df["Category"] == "Furniture"]["Profit"]
category2 = df[df["Category"] == "Technology"]["Profit"]

t_stat, p_value = ttest_ind(category1, category2, equal_var=False)

print("INDEPENDENT T-TEST")
print("=" * 40)
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Significant difference in average profit.")
else:
    print("Result: No significant difference in average profit.")
```

```python
# Chi-Square Test
# Relationship between Category and Region

from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(df["Category"], df["Region"])

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("CHI-SQUARE TEST")
print("=" * 40)
print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)

if p_value < 0.05:
    print("Result: Category and Region have a significant relationship.")
else:
    print("Result: No significant relationship between Category and Region.")
```

```python
if p_value < 0.05:
    print("Result: Significant difference in Profit between Consumer and Corporate segments.")
else:
    print("Result: No significant difference in Profit between Consumer and Corporate segments.")
```

```python
# 95% Confidence Interval for Average Sales

from scipy import stats
import numpy as np

sales = df["Sales"].dropna()

mean_sales = sales.mean()
standard_error = stats.sem(sales)

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(sales) - 1,
    loc=mean_sales,
    scale=standard_error
)

print("95% CONFIDENCE INTERVAL FOR SALES")
print("=" * 40)
print("Mean Sales:", mean_sales)
print("Lower Bound:", confidence_interval[0])
print("Upper Bound:", confidence_interval[1])
```

## Statistical Analysis - Key Insights

1. Sales and Profit were analyzed using mean, median, mode, standard deviation, and skewness.

2. The statistical summary helped identify the central tendency and variation in Sales, Quantity, Discount, and Profit.

3. The skewness analysis shows the distribution pattern of Sales and Profit.

4. The t-test was used to check whether the average Profit differs significantly between Furniture and Technology categories.
