```python
import pandas as pd
```

```python
df=pd.read_csv("data/superstore sales dataset.csv")
```

```python
df.head()
```

```python
df['Sales'].hist()
```

```python
df.boxplot(column='Sales')
```

```python
df['Category'].value_counts().plot(kind='bar')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df[['Sales','Quantity','Discount','Profit']].corr(),annot=True)
plt.show()
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'],dayfirst=True,errors='coerce')
sales_trend=df.groupby('Order Date')['Sales'].sum()
sales_trend.plot(figsize=(12,5))
plt.title('Sales Trend Over Time')
plt.xlabel('Order Date')
plt.ylabel('Sales')
plt.show()
```

```python
df[['Sales','Quantity','Discount','Profit']].describe()
```

## Key Insights

1. The average Sales value is approximately 229.86, while the median Sales is 54.49. This indicates that some high-value orders increase the average sales.

2. The average Quantity per order is approximately 3.78 units, with a median of 3 units.

3. The average Discount is approximately 15.62%, while the median discount is 20%.

4. The average Profit is approximately 28.66, while the median Profit is 8.67, showing that some high-profit orders increase the average.

5. Profit ranges from -6599.98 to 8399.98, indicating that the dataset contains both significant losses and significant profits.
